```python
import numpy as np
# One-dimensional array
a = np.array([10, 20, 30, 40])
print(a)
# Two-dimensional array
b = np.array([[1, 2, 3],
              [4, 5, 6]])
print(b)
```

```python
import numpy as np
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(a + b)
print(b - a)
print(a * b)
print(b / a)
```

```python
import numpy as np
a = np.array([2, 4, 6])
print(a * 2)
print(a + 5)
print(a / 2)
```

```python
import numpy as np
a = np.array([1, 4, 9, 16])
print(np.sqrt(a))
print(np.power(a, 2))
print(np.exp(a))
print(np.log(a))
```

```python
import numpy as np
data = np.array([10, 20, 30, 40, 50])
print(np.sum(data))
print(np.mean(data))
print(np.min(data))
print(np.max(data))
print(np.std(data))
```
